```python
import cv2
import matplotlib.pyplot as plt
from google.colab import files
from IPython.display import HTML, display
from base64 import b64encode

# Upload a video
uploaded = files.upload()
filename = next(iter(uploaded))

# Read video and get its properties
cap = cv2.VideoCapture(filename)

if not cap.isOpened():
    print("Error: Unable to open video!")
else:
    fps = cap.get(cv2.CAP_PROP_FPS)
    width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
    height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

    print("Original FPS:", fps)
    print("Video Width:", width)
    print("Video Height:", height)

    cap.release()

    # Function to change video playback speed
    def change_speed(input_file, output_file, speed):
        cap = cv2.VideoCapture(input_file)

        if not cap.isOpened():
            print("Error: Unable to open video!")
            return

        original_fps = cap.get(cv2.CAP_PROP_FPS)

        if original_fps <= 0:
            original_fps = 25

        new_fps = original_fps * speed

        fourcc = cv2.VideoWriter_fourcc(*'mp4v')
        out = cv2.VideoWriter(
            output_file, fourcc, new_fps, (width, height)
        )

        while True:
            ret, frame = cap.read()

            if not ret:
                break

            out.write(frame)

        cap.release()
        out.release()

    # Create slow-motion and fast-motion videos
    change_speed(filename, "slow_motion.mp4", 0.5)
    change_speed(filename, "fast_motion.mp4", 2.0)

    print("Slow-motion and fast-motion videos created successfully!")

    # Display a video in Google Colab
    def display_video(video_path):
        with open(video_path, "rb") as f:
            video_data = b64encode(f.read()).decode()

        display(HTML(f'''
            <video width="480" controls>
                <source src="data:video/mp4;base64,{video_data}"
                        type="video/mp4">
            </video>
        '''))

    print("Slow Motion Video:")
    display_video("slow_motion.mp4")

    print("Fast Motion Video:")
    display_video("fast_motion.mp4")

    # Download the output videos
    files.download("slow_motion.mp4")
    files.download("fast_motion.mp4")
```
