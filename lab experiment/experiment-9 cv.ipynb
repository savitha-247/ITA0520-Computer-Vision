```python
import cv2
import matplotlib.pyplot as plt
from google.colab import files

# Upload an image
uploaded = files.upload()
filename = next(iter(uploaded))

# Read the image
img = cv2.imread(filename)

if img is None:
    print("Error: Image not found!")
else:
    # Convert BGR to RGB
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    # Resize to a bigger size
    bigger = cv2.resize(
        img, None, fx=2, fy=2,
        interpolation=cv2.INTER_LINEAR
    )

    # Resize to a smaller size
    smaller = cv2.resize(
        img, None, fx=0.5, fy=0.5,
        interpolation=cv2.INTER_AREA
    )

    # Display all images
    plt.figure(figsize=(12, 5))

    plt.subplot(1, 3, 1)
    plt.imshow(img)
    plt.title("Original Image")
    plt.axis("off")

    plt.subplot(1, 3, 2)
    plt.imshow(bigger)
    plt.title("Bigger Image (2x)")
    plt.axis("off")

    plt.subplot(1, 3, 3)
    plt.imshow(smaller)
    plt.title("Smaller Image (0.5x)")
    plt.axis("off")

    plt.tight_layout()
    plt.show()

    # Save the resized images
    cv2.imwrite(
        "bigger_image.jpg",
        cv2.cvtColor(bigger, cv2.COLOR_RGB2BGR)
    )
    cv2.imwrite(
        "smaller_image.jpg",
        cv2.cvtColor(smaller, cv2.COLOR_RGB2BGR)
    )

    # Download the output images
    files.download("bigger_image.jpg")
    files.download("smaller_image.jpg")
```
