```python
import cv2
import matplotlib.pyplot as plt
from google.colab import files

# Upload an image
uploaded = files.upload()
filename = next(iter(uploaded))

# Read the source image
img = cv2.imread(filename)

if img is None:
    print("Error: Image not found!")
else:
    # Get image dimensions
    height, width = img.shape[:2]

    # Define ROI coordinates
    x1 = width // 4
    y1 = height // 4
    x2 = 3 * width // 4
    y2 = 3 * height // 4

    # Crop the selected region
    roi = img[y1:y2, x1:x2]

    # Copy the original image
    copied_img = img.copy()

    # Paste the ROI into the top-left corner
    roi_h, roi_w = roi.shape[:2]
    copied_img[0:roi_h, 0:roi_w] = roi.copy()

    # Convert BGR to RGB for display
    original_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    roi_rgb = cv2.cvtColor(roi, cv2.COLOR_BGR2RGB)
    pasted_rgb = cv2.cvtColor(copied_img, cv2.COLOR_BGR2RGB)

    # Display results
    plt.figure(figsize=(12, 5))

    plt.subplot(1, 3, 1)
    plt.imshow(original_rgb)
    plt.title("Original Image")
    plt.axis("off")

    plt.subplot(1, 3, 2)
    plt.imshow(roi_rgb)
    plt.title("Cropped ROI")
    plt.axis("off")

    plt.subplot(1, 3, 3)
    plt.imshow(pasted_rgb)
    plt.title("ROI Pasted Image")
    plt.axis("off")

    plt.tight_layout()
    plt.show()

    # Save output images
    cv2.imwrite("cropped_roi.jpg", roi)
    cv2.imwrite("pasted_roi_image.jpg", copied_img)

    # Download output images
    files.download("cropped_roi.jpg")
    files.download("pasted_roi_image.jpg")
```
